```python
# 1. Install the correct new package
!pip install -q -U google-genai langchain

# 2. Use the new import statement
from google import genai
from google.colab import userdata

# Fetch your API key from Google Colab's secrets manager
key = userdata.get("Gemini_API_Key")

# 3. Initialize the Client (Configuration has changed in the new SDK)
client = genai.Client(api_key=key)

# Define your prompt
prompt_text = "give me stratergies of clash of clans to attack on th14"

# 4. Generate the response using the client.models method
response = client.models.generate_content(
    model='gemini-3.5-flash',
    contents=prompt_text
)

# Print the result
print(response.text)
```
