```python
!pip install -q -U openai

from openai import OpenAI
from google.colab import userdata

# Fetch your OpenRouter key from Colab secrets
key = userdata.get("Openai")

# Initialize client pointing to OpenRouter
client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=key
)

prompt_text = "Do you know about Clash of Clans"

response = client.chat.completions.create(
    model="openrouter/free",
    messages=[
        {"role": "user", "content": prompt_text}
    ]
)

# Print which specific model was selected by the free router
print(f"Model used: {response.model}")
print("-" * 40)

# Print the result
print(response.choices[0].message.content)
```
